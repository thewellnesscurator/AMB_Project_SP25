#### **Addressing null/missing values**: 

**Determining best method for replacement**
The first step will be to identify how to best address any null values in the dataframes.

**Note**: For these datasets, the null value is in the "*Caffeine*" column of the *SBUX* dataframe. While the plan is to *delete* this column, below is an example of how to address null values for demonstrative purposes

**Method 1:  Converting null/missing values to *zero*:**

*Note*: If the null/missing value represents unknown data or data that is truly missing, converting to zero may contribute to misleading results. It may be better to drop these rows/columns or use statistical methods.

Teas in the SBUX DataFrame were noted to have a non-numeric, variable ("Varies/varies") caffeine quantities and were also converted to zero.

```python
# one missing value was identified in the SBUX dataset in a column which will later be deleted.
# for this dataset type, null value could be changed to 0 as per below, but I would have personally replaced missing values with the mean of similar beverages using the following code:

#verify data type
print(sbux["Caffeine (mg)"].dtype)
#change "Varies/ varies" to a null value
sbux["Caffeine (mg)"] = sbux["Caffeine (mg)"].replace("Varies", 0)
sbux["Caffeine (mg)"] = sbux["Caffeine (mg)"].replace("varies", 0)
#convert null value to 0
sbux["Caffeine (mg)"] = sbux["Caffeine (mg)"].fillna(0)
#verify conversion / check for null values in this column
print(sbux["Caffeine (mg)"].isnull().sum())
```

**Method 2: Converting null/missing values value to *mean* (example of a statistical method)**:

For the scenario where a distinction between the "Iced Brewed Coffee (with Milk & Classic Syrup) with non-null values and the one with a null value was the type of milk (which does not contain caffeine), using an average of similar drinks could be a better replacement but does not account for variation in cup size. 

Again, this column will be deleted for the overall project.

*Note*: Data type and distribution are helpful in determining which statistical method to use.

For *numeric/quantitative data*, 
- **Mean**: best used when data is normally distributed (think bell curve) and there are no extrene outliers
- **Median**: best used when data either skews to the left or right or contains outliers
- **Mode**: helpful in show the most frequent occurences 

For *catgorical/qualitative data*,
- **Mode**: helpful in showing most common value

```python
#convert "Caffeine (mg)" column from object to integer
sbux["Caffeine (mg)"] = sbux["Caffeine (mg)"].astype(int)
#verify new column data type
print(sbux["Caffeine (mg)"].dtype)
```

```python
#determine mean "Caffeine (mg)" for Iced Brewed Coffee (With Milk & Classic Syrup)
similar_rows = sbux.loc["Iced Brewed Coffee (With Milk & Classic Syrup)"]
mean_caffeine = similar_rows['Caffeine (mg)'].mean()

print (mean_caffeine)
```

### Conclusions, Limitations, & Next Steps

#### **Conclusions**

*In comparing calories from Starbucks (SBUX) to McDonald’s (MCD)*:
- For hot and iced coffee, the addition of “classic syrup“ at SBUX increases the calorie count by > 6 to 12 fold.
- From mocha, calories from MCD exceed SBUX. There is a nearly two fold increase in comparison to  nonfat options.
- For frappuccino, data is generally variable. There appears to be less calories at SBUX than at MCD (Frappes). This trend is noted without regard to flavor.
- For vanilla lattes, the use of MCD's sugar-free French vanilla syrup is associated with lower calories than other varieties of vanilla lattes. This improvement in calories trends in the other direction when changing to regular syrup.
- For hot chocolates, the lowest calotirs were seen using nonfat milk without whipped cream at SBUX. This trends upwards with other SBUX hot chocolate or with any MCD options.
- For smoothies, SBUX exceeds the calories of MCDs for similar sizes. However, MCD  offers the largest size at 22 ounces which has the highest calorie content.

*In evaluating macronutrients (macros) by beverage type*:
- For caffé latte, there is a notably high carb content which reflects sugars (> fiber intake). There is a substantial increase in fat intake when switching from non-fat to  2%. Protein content increases with increase in size.
- These findings are similar for caffé mocha as a relates to fats and carbs and for French vanilla lattes show a similar trend as it relates to fats, cards, and protein.
- For Frappuccinoss, SBUX is high in carbs in comparison to MCDs. Whereas MCD exceeds SBUX in percent calories from fat/saturated fat. For carbs, we see a reflection of high sugar and minimal fiber.
- For French vanilla lattes, non-fat/skinny options have substantially less fat and less saturated fat. However, they are not associated with having less carb or sugar content.
-For hot chocolates, high quantities of carbs from sugars (not fiber) in SBUX options are  similarly reflected at MCDs. MCD's hot chocolate is  notable for high fat/saturated fat content.
- Starbucks smoothies are high in carbs from sugar (not fiber). However, fat and saturated fat content is not particularly high.

#### **Limitations**
- A few inherent limitations of the McDonald’s and Starbucks data sets that were utilized for this project were noted. 
- Some beverage items are not labeled with corresponding sizes. 
- Some items do not reflect which type of milk, if any, is included into the beverage.
- There appears to be a discrepancy between what would be the reported amount of calories from carbs vs the anticipated amount based on standard calculations.
- Addressing null/missing values with zero method or with statistical methods, such as mean, median, or mode continues to be associated with the potential for skewing data.

#### **Next Steps**
- Findings highlightes that beverage items from quick service restaurants tend to be
*calorie dense*, i.e. high in saturated fats (fats) and/or added sugars (carbohydrates) with insufficient quantities of essentials like fiber and protein. 
- Similar analysis method can be applied to a variety of food items to obtain similar information and  guidance. 
- This may be helpful in delineating which foods would, alternatively, be considered to be *nutrient dense* (> calorie dense).
